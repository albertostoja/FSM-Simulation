```python
# 1. Load both dataframes
df = pd.read_csv('fsm_fiber_circle_0.005_1Hz.csv')

# 2. Slice the data (every 10th point)
data_sliceI1_df = df['I1'][::50].values
data_sliceI2_df = df['I2'][::50].values

# 3. Create corresponding time/index axes
t_df = np.arange(len(data_sliceI1_df))

# 4. Plotting
plt.figure(figsize=(8, 6))

# Plot first dataset with 'PuBu' colormap
sc1 = plt.scatter(
    data_sliceI1_df,
    data_sliceI2_df,
    c=t_df,
    cmap='PuBu',
    alpha=0.8,
    label='Circle Data',
)

plt.xlabel('I1')
plt.ylabel('I2')
plt.title('Overlay: I1 vs I2 Over Time')

# Add colorbar for the first dataset
plt.colorbar(sc1, label='Time Index (df)')

plt.grid(False)
plt.axis('equal')  # Ensures equal scaling on both axes
plt.tight_layout()
plt.show()
```

```python
# 1. Load both dataframes
df = pd.read_csv('fsm_fiber_circle_0.005_1Hz.csv')
df1 = pd.read_csv('fsm_fiberlaser_basis2.csv')

# 2. Slice the data (every 50th point)
data_sliceI1_df = df['I1'][10000:15000:5].values
data_sliceI2_df = df['I2'][10000:15000:5].values

data_sliceI1_df1 = df1['I1'][::50].values
data_sliceI2_df1 = df1['I2'][::50].values

# 3. Create corresponding time axes in seconds
fs = 50000  # Sampling frequency in Hz
step = 50  # Slicing step
dt = step / fs  # Time increment per slice step (0.001 s)

t_df = np.arange(len(data_sliceI1_df)) * dt
t_df1 = np.arange(len(data_sliceI1_df1)) * dt

# 4. Plotting with swapped axes, larger fonts, and no title
plt.figure(figsize=(9, 7))

# Plot first dataset with 'PuBu' colormap (X: I2, Y: I1)
sc1 = plt.scatter(
    data_sliceI2_df,
    data_sliceI1_df,
    c=t_df,
    cmap='PuBu',
    alpha=0.8,
    label='Circle Data',
)

# Plot second dataset on top with a different colormap ('OrRd') (X: I2, Y: I1)
sc2 = plt.scatter(
    data_sliceI2_df1,
    data_sliceI1_df1,
    c=t_df1,
    cmap='OrRd',
    alpha=0.8,
    label='Basis Data',
)

# Swapped axis labels with larger fonts
plt.xlabel('Horizontal Polarization Intensity ($I_2$)', fontsize=16)
plt.ylabel('Vertical Polarization Intensity ($I_1$)', fontsize=16)

# Axis ticks font size
plt.xticks(fontsize=14)
plt.yticks(fontsize=14)

# Legend font size
plt.legend(fontsize=14)

# Add colorbar with increased font size
cbar = plt.colorbar(sc1)
cbar.set_label('Time (seconds)', fontsize=14)
cbar.ax.tick_params(labelsize=13)

plt.grid(False)
plt.axis('equal')  # Ensures equal scaling on both axes
plt.tight_layout()
plt.show()
```
